```python
df.drop_duplicates(inplace=True)
```

```python
df=df.apply(lambda x:x.str.strip() if x.dtype=='object'else x)
```

```python
df["Usage Date"]=pd.to_datetime(df["Usage Date"],errors="coerce")
```

```python
numeric_cols = ['Daily Usage (Minutes)', 'Satisfaction Score', 'Response Quality Rating', 'Accuracy Rating']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    print(df[numeric_cols].dtypes)
```

```python
df['Month'] = df['Usage Date'].dt.month_name()
df['Year'] = df['Usage Date'].dt.year
df['Weekday'] = df['Usage Date'].dt.day_name()

df[['Usage Date', 'Month', 'Year', 'Weekday']].head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
df['AI / LLM Name'] = df['AI / LLM Name'].str.title()
```

```python
df.drop_duplicates(inplace=True)
```

```python
print("Unique Countries:", df['Country'].unique())
print("Unique Devices:", df['Device'].unique())
```

```python
print("Top 5 Usage Outliers:\n", df['Daily Usage (Minutes)'].nlargest(5))
```

```python
if 'Age' in df.columns:
    df = df[(df['Age'] >= 10) & (df['Age'] <= 90)]
```

```python
print("\nFinal Data Types:\n", df.dtypes)
```

```python
df.to_csv("Ready_llm.csv",index=False, encoding='utf-8-sig')
```
